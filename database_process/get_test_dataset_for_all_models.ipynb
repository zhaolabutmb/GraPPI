### Check if mutants are in our mutant dataset

```python
# Function to normalize mutations: split by ', ', strip, sort, join
def normalize_mutations(mut_str):
    if pd.isna(mut_str):
        return mut_str
    muts = [m.strip() for m in mut_str.split(', ')]
    return ', '.join(sorted(muts))

# Apply normalization to both dataframes
SKEMPI_test_df['Mutations_norm'] = SKEMPI_test_df['Mutations'].apply(normalize_mutations)
mut_df['Mutations_norm'] = mut_df['Mutations'].apply(normalize_mutations)

# Now merge on 'PDB' and 'Mutations_norm'
common_rows = pd.merge(SKEMPI_test_df[['PDB', 'Mutations_norm']], mut_df[['PDB', 'Mutations_norm','mut_paths']], on=['PDB', 'Mutations_norm'], how='inner')
print("Common rows after normalization:")
print(common_rows)
print(f"Number of common rows: {len(common_rows)}")

# For different rows
diff_rows = pd.merge(SKEMPI_test_df[['PDB', 'Mutations_norm']], mut_df[['PDB', 'Mutations_norm']], on=['PDB', 'Mutations_norm'], how='left', indicator=True)
diff_rows = diff_rows[diff_rows['_merge'] == 'left_only'].drop(columns=['_merge'])
print("Different rows after normalization:")
print(diff_rows[diff_rows['Mutations_norm'].notna()])
print(f"Number of different rows: {len(diff_rows)}")
```

# Split internal testset

```python
import sys
sys.path.append("..")
from utils.Training_modules.common_utils import (
    load_sttgs_from_dir, filter_samples_finetune,get_sample_type,
    stratified_train_test_split, save_split_info
)
from collections import defaultdict
```

```python
pdb_dir = "../../GraPPI_data/curated_db/all_sthg_esm_8A/"
all_sttgs = load_sttgs_from_dir(pdb_dir)
all_sttgs = filter_samples_finetune(all_sttgs, min_nodes=5)
# Convert to samples
all_samples = []
for name, st in all_sttgs.items():
	graph = st.protein_graph
	affinity = graph.y.item() if hasattr(graph.y, 'item') else graph.y
	sample_type = get_sample_type(affinity, 1.1)
	all_samples.append((name, graph, affinity, sample_type))
# Count sample types
type_counts = defaultdict(int)
for s in all_samples:
	type_counts[s[3]] += 1

print(f"\nTotal samples: {len(all_samples)}")
print(f"Premium: {type_counts['premium']}, Golden: {type_counts['golden']}, Negative: {type_counts['negative']}")

if len(all_samples) == 0:
	raise ValueError("No samples loaded.")
```

```python
train_samples, test_samples = stratified_train_test_split(
	all_samples,
	test_ratio=0.2,
	seed=42
)
```

```python
final_save_dir = "./"
save_split_info(train_samples, test_samples, final_save_dir, prefix="internal_test")
```
