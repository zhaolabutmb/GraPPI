# Get test random mut table

```python
import pandas as pd
import json
```

```python
rand_mut_df = pd.read_csv('./saved_tables/rand_mut_table.csv')
with open('../../GraPPI_data/trained_data_ssl_finetune/2layers_10hdim_esm_jkmean/classifier_split_info.json') as f:
    split_info = json.load(f)
```

```python
rand_mut_df = rand_mut_df [['PDB','Ligand Chains','Receptor Chains','new_pdb_name', 'mut_paths']]
```

```python
rand_mut_df
```

```python
# get key value with item == "random_mut" in the test negative sources
test_rand_mut_list = [key for key, value in split_info['test_neg_sources'].items() if value == "random_mut"]
print(len(test_rand_mut_list), test_rand_mut_list[:5])
```

```python
sub_df = rand_mut_df[rand_mut_df['new_pdb_name'].isin(test_rand_mut_list)]
sub_df
```

```python
# change mut_paths name to paths for sub_df
sub_df = sub_df.rename(columns={'mut_paths': 'paths'})
    
```

```python
sub_df
```

```python
sub_df.to_csv('./saved_tables/test_rand_mut_table.csv', index=False)
```

# Get test df for positive

```python
with open('../../GraPPI_data/trained_data_ssl_finetune/2layers_10hdim_esm_jkmean/classifier_split_info.json') as f:
    split_info = json.load(f)
wt_df = pd.read_csv('./saved_tables/wt_table.csv')
dimer_df = pd.read_csv('./saved_tables/dimer_table.csv')
# Combine wild-type and dimer dataframes into a single positive dataframe, if PDB column have repeat rows, just use wt_df's entries
posi_df = pd.concat([wt_df, dimer_df], ignore_index=True)
posi_df = posi_df.drop_duplicates(subset='PDB', keep='first')
len(posi_df)
```

```python
test_pos_name_list = split_info['test_pos_names']
# no matter upper or lower case
test_pos_name_list = [name.upper() for name in test_pos_name_list]
posi_df['PDB'] = posi_df['PDB'].str.upper()
sub_test_pos_df = posi_df[posi_df['PDB'].isin(test_pos_name_list)]
sub_test_pos_df
```

```python
sub_test_pos_df.to_csv('./saved_tables/test_posi_df.csv', index=False)
```
